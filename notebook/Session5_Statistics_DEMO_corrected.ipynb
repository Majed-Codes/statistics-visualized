```python
import pandas as pd
from sklearn.datasets import load_iris

iris = load_iris()

df = pd.DataFrame(iris.data, columns=iris.feature_names)
df["species"] = iris.target
# tail
df.head()
```

**Correction.** `species` is a categorical (nominal) label stored as 0/1/2. It is the target, not a measurement, so all numeric statistics below use only the four measurement columns.

```python
features = df.drop(columns="species")          # the 4 continuous measurements
df["species"] = df["species"].astype("category")  # mark the label as categorical
df.info()
features.describe()
```

# Central Tendency

```python
mean_values = features.mean()
median_values = features.median()

print("Mean:\n", mean_values)
print("Median:\n", median_values)
print("Most frequent species (mode):", df["species"].mode()[0])
```

# Variation Measures

**Note.** pandas divides by *n − 1* (sample variance). The lecture's formula divides by *N* (population): use `ddof=0`.

```python
variance = features.var()          # ddof=1: sample variance
std_dev = features.std()

print("Variance:\n", variance)
print("Standard Deviation:\n", std_dev)
print("Population variance (ddof=0):\n", features.var(ddof=0))
```

# IQR

**Note.** pandas interpolates linearly between order statistics. The lecture finds the median of each half. Both are valid conventions.

```python
Q1 = features.quantile(0.25)
Q3 = features.quantile(0.75)
IQR = Q3 - Q1

print(f"IQR:\n{IQR}")

lower, upper = Q1 - 1.5 * IQR, Q3 + 1.5 * IQR
outliers = ((features < lower) | (features > upper)).sum()
print("Values flagged by the 1.5 x IQR rule:\n", outliers)
```

# Z-score

```python
numeric_df = features   # Correction: was df.select_dtypes(include=np.number), which included species

z_scores = (numeric_df - numeric_df.mean()) / numeric_df.std()
z_scores.head()
```

# Covariance Matrix

```python
cov_matrix = numeric_df.cov()
print(cov_matrix)
```

# Correlation Matrix

```python
corr_matrix = numeric_df.corr()
print(corr_matrix)
```

# PCA Connection

**Correction.** A covariance matrix is symmetric, so use `np.linalg.eigh`: real eigenvalues, orthonormal eigenvectors, returned in ascending order (we reverse them).

```python
# Center data
X = numeric_df - numeric_df.mean()

# Covariance matrix
cov_matrix = np.cov(X.T)

# Eigen decomposition (symmetric matrix)
eigvals, eigvecs = np.linalg.eigh(cov_matrix)
order = np.argsort(eigvals)[::-1]
eigvals, eigvecs = eigvals[order], eigvecs[:, order]

print("Eigenvalues:\n", eigvals)
print("Eigenvectors:\n", eigvecs)
print("Explained variance ratio:\n", eigvals / eigvals.sum())
```
